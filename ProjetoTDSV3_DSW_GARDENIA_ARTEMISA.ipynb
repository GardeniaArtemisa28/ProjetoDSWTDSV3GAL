```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import matplotlib.pyplot as plt
!pip install tabulate #para aparecer a tabela organizada
from tabulate import tabulate

# Ler os arquivos CSV
df1 = pd.read_csv('/content/drive/MyDrive/ProjetoTDSV3_DSW_GARDENIA_ARTEMISA/TABELA 1 AGRUPADA.csv', encoding='latin1', delimiter=';')

# Remover linhas com NaN nas colunas "Idade" e "Ano de Nascimento" e criar uma cópia explícita
df1_cleaned = df1.dropna(subset=['Idade', 'Ano de Nascimento']).copy()

# Converter as colunas para inteiro (para não aparecer o ".0" no final )
df1_cleaned['Idade'] = df1_cleaned['Idade'].astype(int)
df1_cleaned['Ano de Nascimento'] = df1_cleaned['Ano de Nascimento'].astype(int)

# Exibir a tabela de forma "bonita"
print(tabulate(df1_cleaned, headers='keys', tablefmt='pretty', showindex=False))

# Contar a frequência de cada impacto emocional
impactos = df1_cleaned['Impacto Emocional'].str.split(', ', expand=True).stack().value_counts()

# Plotar o gráfico de pizza
plt.figure(figsize=(8, 8))
impactos.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['lightcoral', 'lightgreen', 'lightskyblue', 'gold' , 'red', 'purple'])
plt.title('Proporção dos Impactos Emocionais')
plt.ylabel('')  # Remove o rótulo do eixo Y
plt.show()

# Contar a frequência de cada ameaça cibernética
ameacas = df1_cleaned['Ameaças Cibernéticas Sucetíveis'].str.split(', ', expand=True).stack().value_counts()

# Plotar o gráfico de barras
plt.figure(figsize=(10, 6))
ameacas.plot(kind='bar', color='skyblue')
plt.title('Frequência de Ameaças Cibernéticas')
plt.xlabel('Tipo de Ameaça')
plt.ylabel('Frequência')
plt.xticks(rotation=45, ha='right')
plt.show()

# Salvar o DataFrame como um arquivo CSV
df1_cleaned.to_csv('TABELA_1_AGRUPADA_CLEANED.csv', index=False, encoding='latin1', sep=';')
```
